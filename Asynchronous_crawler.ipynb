```python
import asyncio
import aiohttp
import time

urls = [
    "https://example.com",
    "https://www.python.org",
    "https://www.google.com"
]

# Sequential crawler
def sequential_crawler():
    import requests

    start = time.time()

    for url in urls:
        try:
            response = requests.get(url, timeout=5)
            print("Sequential:", url, response.status_code)
        except Exception as e:
            print("Sequential Error:", url)

    return time.time() - start


# Asynchronous crawler with retries
async def fetch(session, url, retries=3):
    for attempt in range(retries):
        try:
            async with session.get(url, timeout=5) as response:
                await response.text()
                print("Async:", url, response.status)
                return True

        except Exception:
            if attempt < retries - 1:
                await asyncio.sleep(1)

    print("Failed:", url)
    return False


async def asynchronous_crawler():
    start = time.time()

    async with aiohttp.ClientSession() as session:
        tasks = [fetch(session, url) for url in urls]
        results = await asyncio.gather(*tasks)

    return time.time() - start


# Run in Jupyter / Colab
seq_time = sequential_crawler()

async_time = await asynchronous_crawler()

print("\n----- RESULT -----")
print("Sequential Time :", round(seq_time, 2), "seconds")
print("Async Time      :", round(async_time, 2), "seconds")
```
